```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Load data
df = pd.read_csv("kc_house_data_original.csv")
```

```python
df["date"] = pd.to_datetime(df["date"]) # convert to datetime since its string 


#create new DF to capture monthly sales 
monthly_sales = ( df.groupby(df["date"].dt.to_period("M")).size()
)

monthly_sales.index = monthly_sales.index.to_timestamp() 

plt.plot(monthly_sales)
plt.show()
```
