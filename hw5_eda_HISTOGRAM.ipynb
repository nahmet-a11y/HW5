```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Load data
df = pd.read_csv("kc_house_data_original.csv")
grid = sns.FacetGrid(df, col="waterfront", margin_titles=True)

grid.map(plt.hist, "price", bins=np.linspace(0, df["price"].max(), 10))

plt.show()
```
